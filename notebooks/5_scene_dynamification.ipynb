## Convex Decomposition


```python
import coacd
import trimesh

mesh = trimesh.load("/home/jvermandere/projects/DRM/_input/HalfChair.obj", force="mesh")
mesh = coacd.Mesh(mesh.vertices, mesh.faces)
parts = coacd.run_coacd(mesh) # a list of convex hulls.

# Or use real metric mode (threshold in meters)
#parts = coacd.run_coacd(mesh, threshold=0.01, real_metric=True)
```

```python
scene = trimesh.Scene()
for vs, fs in parts:
    hull = trimesh.Trimesh(vertices=vs, faces=fs)
    hull.visual.face_colors = trimesh.visual.random_color()
    scene.add_geometry(hull)

scene.show()
```
